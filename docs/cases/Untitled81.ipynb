Lehman Brothers filed for bankruptcy on September 15, 2008 despite reporting capital ratios that appeared comparable to its peers.

This case examines a central paradox in financial risk:

> A firm can appear well-capitalized — yet fail rapidly due to a collapse in confidence and liquidity.

## Data Overview

The following data track:

- Gross leverage (assets / equity)  
- Net leverage  
- Capital proxy (equity-to-assets)  
- Tier 1 capital (where available)  

```python
import numpy as np
import matplotlib.pyplot as plt

labels = ["2003","2004","2005","2006","2007","2008 Q2","2008 Q3"]
leverage = np.array([23.7, 23.9, 24.4, 26.2, 30.7, 24.3, 21.1])
net_leverage = np.array([15.3, 13.9, 13.6, 14.5, 16.1, 12.1, 10.6])
capital_proxy = 100 / leverage
tier1 = np.array([np.nan, np.nan, np.nan, np.nan, np.nan, 11.1, 11.0])
```

## Q1 — Capital Measures

We compare:

- Regulatory Tier 1 capital  
- A simple capital proxy (equity-to-assets) derived from leverage  

These measures can behave differently — especially for trading-heavy institutions.

```python
fig, ax = plt.subplots(figsize=(10,6))

ax.plot(labels, capital_proxy, marker='^', label='Capital Proxy (%)')
ax.plot(labels, tier1, marker='o', label='Tier 1 (%)')

ax.set_title("Lehman: Tier 1 vs Capital Proxy")
ax.set_ylabel("Percent (%)")
ax.grid(True)
ax.legend()

plt.show()
```

## Q2 — Liquidity vs Solvency

A firm can be:

- **Solvent** (assets > liabilities)  
- Yet still **fail due to liquidity collapse**

Lehman’s failure provides a clear example of this distinction.

### Conceptual Understanding

- Solvency = long-term balance sheet condition  
- Liquidity = short-term funding ability  

In September 2008, markets focused primarily on liquidity, not capital ratios.
