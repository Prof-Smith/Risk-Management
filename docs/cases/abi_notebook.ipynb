## Stage 1: Introducing ABI and Basel I

## Introducing Atlas Bank International (ABI)

ABI is a globally active, aggressively trading universal bank with major exposures in structured credit, equities, commodities, and rates. Its business model before the financial crisis relied heavily on:

- large proprietary trading desks,
- warehousing mezzanine and equity tranches of CDOs,
- short-term repo funding,
- heavy use of off–balance sheet vehicles.

Basel I: Foundations and FlawsBasel I (1988) introduced the first global framework for bank capital, requiring total capital of at least 8% of Risk-Weighted Assets (RWA). However:

- Risk weights were simple and coarse.
- All corporate loans received 100% risk weight regardless of credit quality.
- Off–balance sheet exposures were underestimated. ABI exploited these weaknesses through regulatory arbitrage, shifting into higher-yielding but riskier assets without requiring more capital.

```python
import plotly.graph_objects as go
fig = go.Figure(go.Scatter(x=[1988,1996,2007,2010], y=[1,2,3,4],    mode='lines+markers+text', text=['Basel I','Amend','Basel II','Basel III'], textposition='top center'))
fig.update_layout(plot_bgcolor='white', paper_bgcolor='white', title='Basel Timeline')
fig.show()
```

## Stage 2: Basel II - Internal Models and Hidden Vulnerabilities

Basel II introduced the Internal Ratings-Based (IRB) approach allowing banks like ABI to estimate PD, LGD, EAD. Calm period data from 2003-2006 led to underestimated risk.

- Procyclicality Low volatility reduced VaR despite growing exposures.
- Supervisory IssuesBacktesting rarely failed; stress testing was inconsistent.
- ABI Pre-crisisTrading book expanded; VaR ~42M.

Under Basel II, Atlas Bank International (ABI) utilized the Internal Ratings-Based (IRB) approach, which allowed them to estimate key risk parameters like Probability of Default (PD), Loss Given Default (LGD), and Exposure At Default (EAD. However, ABI's VaR (Value at Risk) calculations were flawed. The estimations were based on calm period data from 2003-2006, leading to an underestimation of actual risk. This also contributed to procyclicality, where low volatility reduced VaR despite increasing exposures. Furthermore, supervisory issues were present, with backtesting rarely failing and stress testing being inconsistent. Prior to the crisis, ABI's trading book expanded significantly, and its VaR was approximately $42 million.


```python
import plotly.graph_objects as go
years=['2004','2005','2006','2007']
var_vals=[35,40,38,42]
fig=go.Figure(go.Scatter(x=years,y=var_vals,mode='lines+markers'))
fig.update_layout(title='ABI VaR under Basel II',plot_bgcolor='white',paper_bgcolor='white')
fig.show()
```

## Stage 3: Crisis Hits ABI

- Full-Scale Market ShockWhen markets turned in 2007-08, volatility surged, credit spreads widened, and ABI's trading positions suffered severe losses far beyond VaR predictions.
- Funding liquidity evaporated; repo haircuts jumped.
- Margin calls forced asset fire-sales, amplifying losses.
- Structured credit, rates, equities, and commodities desks all posted deep losses.

The 'ABI: VaR vs. Realized Losses' plot illustrates the stark difference between Atlas Bank International's (ABI) Value at Risk (VaR) estimates and its actual realized losses during the financial crisis. As the plot shows, ABI's VaR values, represented by the blue line, were relatively low and slowly increasing from 2006 to 2008 (38M, 42M, 50M respectively). In contrast, the realized losses, shown by the orange line, were significantly higher and rapidly escalated during the crisis, going from 200M in 2006 to 4500M in 2008. This indicates that ABI's VaR models severely underestimated the potential for losses, particularly the tail risk, during periods of market stress, leading to massive unexpected financial damage.

```python
import plotly.graph_objects as go
years=['2006','2007','2008']
var_vals=[38,42,50]
real_losses=[abs(-200),abs(-800),abs(-4500)]
fig=go.Figure()
fig.add_trace(go.Scatter(x=years,y=var_vals,name='VaR'))
fig.add_trace(go.Scatter(x=years,y=real_losses,name='Realized Losses'))
fig.update_layout(title='ABI: VaR vs Realized Losses',plot_bgcolor='white',paper_bgcolor='white', yaxis_title='Value ($M)')
fig.show()
```

## Stage 4: Basel II.5

- Emergency Market Risk Reinforcement: Basel II.5 was created to address clear modeling failures exposed during the crisis. ABI's trading losses revealed VaR severely underestimated tail risk.
- Stressed VaR (sVaR): Uses a fixed 250-day crisis window, forcing capital to reflect stressed conditions.
- Incremental Risk Charge (IRC): Captures credit migration and default risk for traded credit products.
- Comprehensive Risk Measure (CRM): Applied to correlation trading positions which suffered significant losses during the crisis.
- ABI ImpactTrading-book capital requirements exceeded $3B under Basel II.5.

The 'ABI VaR vs Stressed VaR' plot illustrates the difference between ABI's traditional VaR and the newly introduced Stressed VaR (sVaR) under Basel II.5. Basel II.5 was created to address the clear modeling failures exposed during the crisis, where ABI's trading losses revealed that VaR severely underestimated tail risk. Stressed VaR uses a fixed 250-day crisis window, which forces capital requirements to reflect stressed conditions, typically resulting in higher values compared to regular VaR. For example, the plot shows that while VaR was relatively low (38M, 42M, 50M for 2006, 2007, 2008 respectively), Stressed VaR was significantly higher (120M, 140M, 160M for the same years), indicating a more conservative and realistic assessment of risk during turbulent periods.

```python
import plotly.graph_objects as go
years=['2006','2007','2008']
var_vals=[38,42,50]
svar_vals=[120,140,160]
fig=go.Figure()
fig.add_trace(go.Bar(x=years,y=var_vals,name='VaR'))
fig.add_trace(go.Bar(x=years,y=svar_vals,name='Stressed VaR'))
fig.update_layout(barmode='group',title='ABI VaR vs Stressed VaR',plot_bgcolor='white',paper_bgcolor='white')
fig.show()
```

## Stage 5: Basel III

- Full System Rebuild: Basel III strengthened capital quality (CET1), introduced buffers, leverage ratio, LCR, NSFR.
- ET1 and Buffers: Higher standards for equity capital.
- Leverage Ratio: A non-risk-based backstop.
- Liquidity Standards: LCR ensures 30-day stress survival; NSFR ensures stable funding.
- ABI Impact: ABI had to raise CET1, extend funding maturity, increase HQLA.

**Visualizations:**

First, there is a bar chart comparing Basel II and Basel III capital requirements. This plot shows the differences in minimum capital ratios for CET1 (Common Equity Tier 1), Tier 1, and Total Capital under the two regulatory frameworks. As per Basel III, the requirements are higher across the board (CET1: 7%, Tier1: 8.5%, Total: 10.5%) compared to Basel II (CET1: 4.5%, Tier1: 6%, Total: 8%), reflecting the enhanced capital quality and quantity standards introduced by Basel III.

The second plot displays ABI LCR Components. This bar chart shows the values for 'HQLA' (High-Quality Liquid Assets) at 410 and 'Outflows' at 380. This plot visually represents the key components of the Liquidity Coverage Ratio (LCR) for ABI, indicating its ability to cover short-term liquidity needs. A higher HQLA relative to outflows suggests a stronger liquidity position, which is a critical aspect of Basel III's focus on liquidity standards.

```python
import plotly.graph_objects as go
cats=['CET1','Tier1','Total']
b2=[4.5,6,8]
b3=[7,8.5,10.5]
fig=go.Figure()
fig.add_trace(go.Bar(x=cats,y=b2,name='Basel II'))
fig.add_trace(go.Bar(x=cats,y=b3,name='Basel III'))
fig.update_layout(barmode='group',plot_bgcolor='white',paper_bgcolor='white')
fig.show()
```

## Stage 6: Strategic Implications Under Basel III

ABI must restructure: shrink trading book, increase HQLA, extend funding maturity, pivot to fee income.
- Trading Book ReductionHigh capital charges force reduction.
- Funding Structure OptimizationShift to long-term debt.
- Liquidity EnhancementLCR and NSFR reshape liquidity profile.

The visualization in Stage 6 is titled 'ABI Balance Sheet Transformation'. This bar chart illustrates the strategic changes ABI undertook in its balance sheet composition due to the implementation of Basel III. It compares the asset allocation 'Before' and 'After' these changes:

- **Trading Assets:** Reduced from 600 ('Before') to 300 ('After'). This significant reduction reflects the impact of Basel III's higher capital charges on trading book activities, forcing ABI to shrink its exposure in this area.
- **Loans:** Slightly increased from 400 ('Before') to 450 ('After'). This indicates a possible shift towards traditional lending, which might have more stable risk profiles or lower capital requirements compared to trading.
- **HQLA (High-Quality Liquid Assets):** Substantially increased from 100 ('Before') to 300 ('After'). This critical change shows ABI's response to Basel III's new liquidity standards, such as the LCR and NSFR, by increasing its holdings of highly liquid assets to better withstand liquidity shocks.

```python
import plotly.graph_objects as go
labels=['Trading Assets','Loans','HQLA']
before=[600,400,100]
after=[300,450,300]
fig=go.Figure()
fig.add_trace(go.Bar(x=labels,y=before,name='Before'))
fig.add_trace(go.Bar(x=labels,y=after,name='After'))
fig.update_layout(barmode='group',plot_bgcolor='white',paper_bgcolor='white',title='ABI Balance Sheet Transformation')
fig.show()
```

The 'ABI Stability Metrics Summary' plot is a bar chart that displays key stability metrics for Atlas Bank International (ABI). These metrics are crucial under Basel III for assessing a bank's capital adequacy and liquidity:

- **CET1 (Common Equity Tier 1):** With a value of 7, this represents ABI's CET1 capital ratio. Basel III sets a minimum requirement for CET1, and this metric indicates ABI's adherence to that standard.
- **Leverage:** With a value of 5, this refers to ABI's leverage ratio. This non-risk-based backstop limits the build-up of excessive leverage in the banking system.
- **LCR (Liquidity Coverage Ratio):** A value of 110 indicates that ABI can cover 110% of its short-term liquidity needs for a 30-day stress scenario. Basel III introduced the LCR to ensure banks maintain sufficient High-Quality Liquid Assets (HQLA).
